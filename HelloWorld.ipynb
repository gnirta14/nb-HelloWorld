```python
import os
import platform
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from google.colab import drive

goog_drive_home = '/content/drive'
drive.mount(goog_drive_home,force_remount=True)
goog_drive_home = goog_drive_home + '/My Drive'

current_directory = os.getcwd()
print('****',current_directory)

#files = [f for f in os.listdir() if os.path.isfile(f)]
files = [f for f in os.listdir(goog_drive_home)]
for file in files:
    print('**** file',file)

#df = pd.read_csv('/home/tnbt/source/anaconda/notebook/moshTutorial/vgsales.csv')
music_data = pd.read_csv(goog_drive_home + '/music.csv')

# customary to name this X and y
X = music_data.drop(columns=['genre'])
y = music_data['genre']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# new model instance
model= DecisionTreeClassifier()
model.fit(X_train.values,y_train)
predictions = model.predict( X_test.values )

score = accuracy_score(y_test, predictions)
print('\n**** X_train ****\n',X_train,"\n\n**** X_test ****\n", X_test)
score
```
